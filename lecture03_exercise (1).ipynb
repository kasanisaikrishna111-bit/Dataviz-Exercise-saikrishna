```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset: CO2 Emissions by Country 2000-2022
# Source: Our World in Data (https://ourworldindata.org/co2-emissions)
df = pd.read_csv('co2_emissions.csv')
print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

## Task 1 - A line chart showing CO2 per Capita for all countries in the dataset

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#EEEEEE), thinner
- Highlighted country **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country and its story

```python
# Task 1: CO2 per capita for Asian countries

highlight_country = "India"

# Filter the dataset to Asian countries
asia_df = df[df["Region"] == "Asia"].copy()

fig1 = go.Figure()

# Draw all Asian countries in thin grey lines
for country in asia_df["Country"].unique():
    country_data = (
        asia_df[asia_df["Country"] == country]
        .sort_values("Year")
    )

    fig1.add_trace(go.Scatter(
        x=country_data["Year"],
        y=country_data["CO2_per_capita"],
        mode="lines",
        line=dict(color="#E0E0E0", width=1),
        showlegend=False,
        hovertemplate=(
            f"{country}<br>Year: %{{x}}"
            "<br>CO2 per capita: %{y:.2f}<extra></extra>"
        )
    ))

# Select India
india = (
    asia_df[asia_df["Country"] == highlight_country]
    .sort_values("Year")
)

# Highlight India in blue
fig1.add_trace(go.Scatter(
    x=india["Year"],
    y=india["CO2_per_capita"],
    mode="lines",
    line=dict(color="#0077B6", width=3),
    showlegend=False
))

# Label India at the end of the line
last_india = india.iloc[-1]

fig1.add_annotation(
    x=last_india["Year"],
    y=last_india["CO2_per_capita"],
    text="India",
    showarrow=False,
    xshift=25,
    font=dict(color="#0077B6", size=13)
)

# Style the chart
fig1.update_layout(
    title="CO2 emissions per person in Asian countries",
    xaxis_title="Year",
    yaxis_title="CO2 emissions per capita",
    template="plotly_white",
    showlegend=False,
    plot_bgcolor="white",
    paper_bgcolor="white",
    margin=dict(l=50, r=100, t=70, b=50)
)

fig1.update_xaxes(showgrid=False, showline=False)
fig1.update_yaxes(showgrid=True, gridcolor="#EEEEEE", showline=False)

fig1.show()
```

## Task 2 — A line chart showing CO2 emissions over time for all Asian countries in the dataset, with one country highlighted.

**same requirements as Task1**

```python


highlight_country = "India"


asia_df = df[df["Region"] == "Asia"].copy()

fig2 = go.Figure()


for country in asia_df["Country"].unique():
    country_data = (
        asia_df[asia_df["Country"] == country]
        .sort_values("Year")
    )

    fig2.add_trace(go.Scatter(
        x=country_data["Year"],
        y=country_data["CO2_Mt"],
        mode="lines",
        line=dict(color="#D9D9D9", width=1.5),
        showlegend=False,
        hovertemplate=(
            f"{country}<br>Year: %{{x}}"
            "<br>CO2 emissions: %{y:.1f} Mt<extra></extra>"
        )
    ))

# Select India
india = (
    asia_df[asia_df["Country"] == highlight_country]
    .sort_values("Year")
)

# Highlight India in blue
fig2.add_trace(go.Scatter(
    x=india["Year"],
    y=india["CO2_Mt"],
    mode="lines",
    line=dict(color="#0077B6", width=3),
    showlegend=False
))

# Label India at the end of the line
last_india = india.iloc[-1]

fig2.add_annotation(
    x=last_india["Year"],
    y=last_india["CO2_Mt"],
    text="India",
    showarrow=False,
    xshift=25,
    font=dict(color="#0077B6", size=13)
)

# Style the chart
fig2.update_layout(
    title="Total CO2 emissions in Asian countries",
    xaxis_title="Year",
    yaxis_title="CO2 emissions (million tonnes)",
    template="plotly_white",
    showlegend=False,
    plot_bgcolor="white",
    paper_bgcolor="white",
    margin=dict(l=50, r=100, t=70, b=50)
)

fig2.update_xaxes(showgrid=False, showline=False)
fig2.update_yaxes(showgrid=True, gridcolor="#EEEEEE", showline=False)

fig2.show()
```
